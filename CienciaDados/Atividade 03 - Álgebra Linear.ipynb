## 1.
Escreva um programa que cria duas matrizes numpy aleatórias de tamanho 3x3 e realize as seguintes operações (Veja a documentação sobre o módulo random do numpy ou utilize a própria biblioteca random nativa do python para gerar os valores).:

*   *Adição das duas matrizes*
*   *Subtração da primeira matriz pela segunda matriz*
*   *Multiplicação elemento a elemento das duas matrizes*
*   *Multiplicação matricial das duas matrizes (dot)*


```python
import numpy as np

np.random.seed(42)
matriz_a = np.random.randint(1, 10, size=(3, 3))
matriz_b = np.random.randint(1, 10, size=(3, 3))

print("Matriz A:\n", matriz_a)
print("\nMatriz B:\n", matriz_b)
print("\nAdição:\n", matriz_a + matriz_b)
print("\nSubtração (A - B):\n", matriz_a - matriz_b)
print("\nMultiplicação elemento a elemento:\n", matriz_a * matriz_b)
print("\nMultiplicação matricial (dot):\n", np.dot(matriz_a, matriz_b))
```

## 2.
*Crie um programa que gere uma matriz numpy com 20 números inteiros aleatórios no intervalo de 1 a 100. Em seguida, calcule e imprima a média, o desvio padrão, o valor mínimo e o valor máximo dessa matriz.*

```python
import numpy as np

np.random.seed(42)
numeros = np.random.randint(1, 101, size=20)

print("Matriz:", numeros)
print("Média:", np.mean(numeros))
print("Desvio padrão:", np.std(numeros))
print("Valor mínimo:", np.min(numeros))
print("Valor máximo:", np.max(numeros))
```

## 3.
*Crie uma matriz numpy 4x4 com números aleatórios de 1 a 10. Em seguida, imprima:*

* *A primeira linha da matriz*
* *O último elemento da segunda coluna*
* *Um subconjunto da matriz que contenha os elementos das duas primeiras linhas e das duas primeiras colunas.*

```python
import numpy as np

np.random.seed(42)
matriz = np.random.randint(1, 11, size=(4, 4))

print("Matriz:\n", matriz)
print("\nPrimeira linha:", matriz[0])
print("Último elemento da segunda coluna:", matriz[-1, 1])
print("\nSubconjunto 2x2:\n", matriz[:2, :2])
```

## 4.
*Escreva um programa que crie uma matriz numpy 3x3 de uns e outra matriz numpy 3x1 de uns. Em seguida, calcule o produto matricial dessas duas matrizes.*

```python
import numpy as np

matriz_3x3 = np.ones((3, 3), dtype=int)
matriz_3x1 = np.ones((3, 1), dtype=int)

print("Matriz 3x3 de uns:\n", matriz_3x3)
print("\nMatriz 3x1 de uns:\n", matriz_3x1)
print("\nProduto matricial:\n", np.matmul(matriz_3x3, matriz_3x1))
```

## 5.
*Crie uma matriz numpy com valores de 0 a 9. Em seguida, aplique as seguintes funções universais (ufuncs) da NumPy a essa matriz e imprima os resultados:*

* np.sqrt() para calcular a raiz quadrada de cada elemento
* np.exp() para calcular o exponencial de cada elemento
* np.sin() para calcular o seno de cada elemento

```python
import numpy as np

valores = np.arange(10)

print("Valores:", valores)
print("Raiz quadrada:", np.sqrt(valores))
print("Exponencial:", np.exp(valores))
print("Seno:", np.sin(valores))
```
